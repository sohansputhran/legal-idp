Reading FUNSD Dataset from https://guillaumejaume.github.io/FUNSD/

```python
import json
from pathlib import Path
from collections import Counter
```

```python
ann_dir = Path("../data/funsd/training_data/annotations")
label_counter = Counter()
```

```python
for file in ann_dir.glob("*.json"):
    with open(file, encoding="utf-8") as f:
        data = json.load(f)
    for entry in data["form"]:
        label_counter[entry["label"]] += 1
```

```python
total = sum(label_counter.values())
for label, count in label_counter.most_common():
    print(f"{label:10s} {count:6d}  ({count/total:.1%})")
```

```python
print(f"\nTotal entries: {total}")
print(f"Total files: {len(list(ann_dir.glob('*.json')))}")
```
