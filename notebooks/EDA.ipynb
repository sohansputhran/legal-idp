Reading FUNSD Dataset from https://guillaumejaume.github.io/FUNSD/

```python
import json
from pathlib import Path
from collections import Counter

from PIL import Image
```

```python
ann_dir = Path("../data/funsd/training_data/annotations")
img_dir = Path("../data/funsd/training_data/images")

label_counter = Counter()
```

```python
for file in ann_dir.glob("*.json"):
    with open(file, encoding="utf-8") as f:
        data = json.load(f)
    for entry in data["form"]:
        label_counter[entry["label"]] += 1
```

```python
total = sum(label_counter.values())
for label, count in label_counter.most_common():
    print(f"{label:10s} {count:6d}  ({count/total:.1%})")
```

```python
print(f"\nTotal entries: {total}")
print(f"Total files: {len(list(ann_dir.glob('*.json')))}")
```

```python
sample_ann = next(ann_dir.glob("*.json"))
stem = sample_ann.stem  # filename without extension
```

```python
# find matching image (extension may vary)
img_path = next(img_dir.glob(f"{stem}.*"))

img = Image.open(img_path)
width, height = img.size
print(f"Image: {img_path.name}, size: {width}x{height}")
```

```python
with open(sample_ann, encoding="utf-8") as f:
    data = json.load(f)

entry = data["form"][0]
box = entry["box"]
print(f"Sample box: {box}")
print(f"Within bounds: x<=w? {box[2] <= width}, y<=h? {box[3] <= height}")
```
